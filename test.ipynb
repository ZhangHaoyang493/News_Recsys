```python
import torch
import torch.nn as nn

# 替换为您实际的ckpt路径
ckpt_path = '/data2/zhy/News_Recsys/pretrain_embedding/abstract_entity_embedding.ckpt' 
target_row_idx = 2 # 指定要打印的行索引 (从0开始，例如2代表第3行)

try:
    # 加载 checkpoint
    # map_location='cpu' 确保在没有GPU的环境下也能加载
    loaded_data = torch.load(ckpt_path, map_location='cpu')
    
    print(f"Loaded data type: {type(loaded_data)}")

    # 检查类型并打印指定行
    if isinstance(loaded_data, nn.Embedding):
        print("Detected nn.Embedding object.")
        # 获取权重矩阵的指定行
        vector = loaded_data.weight.data[target_row_idx]
        print(f"Vector at index {target_row_idx}: \n{vector}")
        print(f"Vector shape: {vector.shape}")
        
    elif isinstance(loaded_data, torch.Tensor):
        print("Detected torch.Tensor.")
        vector = loaded_data[target_row_idx]
        print(f"Vector at index {target_row_idx}: \n{vector}")
        
    elif isinstance(loaded_data, dict):
        print("Detected state_dict (dict). Keys:", loaded_data.keys())
        # 尝试寻找常见的权重键名
        for key in ['weight', 'embedding.weight', 'embeddings.weight']:
            if key in loaded_data:
                vector = loaded_data[key][target_row_idx]
                print(f"Found '{key}', vector at index {target_row_idx}: \n{vector}")
                break
    else:
        print(f"Unsupported type: {type(loaded_data)}")
        
except FileNotFoundError:
    print(f"File not found: {ckpt_path}")
except IndexError:
    print(f"Index {target_row_idx} is out of bounds.")
except Exception as e:
    print(f"An error occurred: {e}")
```
